```python
vertices = [Vertex(i, (0,0)) for i in range(6)]
edges = [
    Edge(i, vertices[i], vertices[i+1], 'red') 
    for i in range(5)
    ] + [
    Edge(5, vertices[-1], vertices[0], 'red')
        ]

plaquette_e = Plaquette(0, colour='red', edges=edges)
plaquette_v = Plaquette(1, colour='red', vertices=vertices)
```
